```python
import scanpy as sc
import pandas as pd
import numpy as np
```

```python
adata = sc.read('../preprocessing/data/import/Tcell.h5ad')
```

```python
bdata = adata[adata.obs.pheno=='CD8.TemActive']
```

```python
clustering_res=[0.6,0.8,1.0,1.2,1.5,2.0]
[ sc.tl.leiden(bdata, resolution=i, key_added='leiden_'+str(i)) for i in clustering_res ]
```

```python
cdata = adata[adata.obs.pheno.isin(['CD8.TemActive','CD8.Tex'])]
```

```python
cdata.obs = cdata.obs.join(bdata.obs[['leiden_0.6']].rename({'leiden_0.6':'CD69_sub'},axis=1))
```

```python
cdata.obs.CD69_sub = np.where(pd.isna(cdata.obs.CD69_sub),'CD8.Tex',cdata.obs.CD69_sub)
```

```python
tex_markers = ['CD69','HLA-DRA','HLA-DRB1','ICOS','PDCD1','TIGIT',
               'LAG3','CTLA4','HAVCR2','CXCL13','ENTPD1','LAYN','TOX','TOX2']
```

```python
import matplotlib.pyplot as plt
import matplotlib

# Configure matplotlib for editable PDF text
matplotlib.rcParams['pdf.fonttype'] = 42  # TrueType fonts (editable in Illustrator)
matplotlib.rcParams['font.family'] = 'sans-serif'
matplotlib.rcParams['font.sans-serif'] = ['Arial']

# Create and save the plot
fig = sc.pl.matrixplot(cdata, tex_markers, 'CD69_sub', 
                       use_raw=False, swap_axes=False, 
                       standard_scale='var', dendrogram=False, 
                       cmap='Reds', return_fig=True)

fig.savefig('CD8TemCD69_subclus_Tex.pdf', dpi=300)
plt.close()
```
